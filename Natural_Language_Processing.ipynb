```python
def Remove_Html_tag(text):
  pattern=re.compile('<.*?>')
  return pattern.sub(r'',text)
```

```python
exclude=string.punctuation
def Remove_punctutation(text):
  return text.translate(str.maketrans('','',exclude))
```

```python
def Remove_Url(text):
  pattern_ulr=re.compile(r"https*?://\S+|www\.\S+")
  return pattern_ulr.sub(r'',text)
```

```python
def Chat_converstion(text):
  new_chat=[]
  for word in text.split():
    if word.upper() in chat_short_form:
      new_chat.append(chat_short_form[word.upper()])
    else:
      new_chat.append(word)
  return " ".join(new_chat)
```

```python

def Text_correction(text):
  text_blob=TextBlob(text)
  return text_blob.correct().string
```
